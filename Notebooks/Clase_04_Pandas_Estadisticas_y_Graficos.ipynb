## 0) Preparación

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("pandas:", pd.__version__)
```

## 1) Crear un DataFrame
Un DataFrame es una tabla (filas y columnas).

```python
df = pd.DataFrame({
    "id": ["A001", "A002", "A003", "A004"],
    "edad": [18, 20, 19, 21],
    "nota": [15, 17, 14, 18]
})
df
```

## 2) Estadísticas básicas
- `mean()` promedio
- `var()` varianza
- `median()` mediana
- `describe()` resumen

```python
print("Promedio nota:", df["nota"].mean())
print("Varianza nota (poblacional):", df["nota"].var(ddof=0))
print(df.describe(include="all"))
```

## 3) Filtrado y nuevas columnas

```python
aprobados = df[df["nota"] >= 15]
print("Aprobados:")
display(aprobados)

df["nota_norm"] = (df["nota"] - df["nota"].min()) / (df["nota"].max() - df["nota"].min())
df
```

## 4) Gráficos

```python
plt.figure()
plt.hist(df["nota"], bins=4)
plt.title("Histograma de notas")
plt.xlabel("nota")
plt.ylabel("frecuencia")
plt.show()
```
