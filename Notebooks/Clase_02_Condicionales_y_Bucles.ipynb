## 1) Condicionales (if / elif / else)
Sirven para ejecutar código dependiendo de una condición booleana (`True`/`False`).

```python
x = 7
if x > 10:
    print("x es mayor que 10")
elif x == 10:
    print("x es igual a 10")
else:
    print("x es menor que 10")
```

### Operadores comunes
- Comparación: `==`, `!=`, `<`, `<=`, `>`, `>=`
- Lógicos: `and`, `or`, `not`
- Pertenencia: `in`

```python
a = 5
b = 12
print(a < b and b < 20)
print("py" in "python")
```

## 2) Bucle for
Recorre elementos de una colección o una secuencia con `range`.

```python
# range(n) -> 0..n-1
suma = 0
for i in range(5):
    suma += i
print("suma 0..4 =", suma)
```

## 3) Bucle while
Repite mientras la condición sea verdadera.

```python
n = 5
fact = 1
while n > 1:
    fact *= n
    n -= 1
print("5! =", fact)
```
