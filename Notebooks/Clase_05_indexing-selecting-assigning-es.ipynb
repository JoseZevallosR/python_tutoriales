## 5.

Selecciona los registros con etiquetas de índice `1`, `2`, `3`, `5` y `8`, y asigna el resultado a la variable `sample_reviews`.

## 6.

Crea una variable `df` que contenga las columnas `country`, `province`, `region_1` y `region_2` de los registros con etiquetas de índice `0`, `1`, `10` y `100`. En otras palabras, genera el siguiente `DataFrame`:

## 7.

Crea una variable `df` que contenga las columnas `country` y `variety` de los primeros 100 registros. 

Pista: puedes usar `loc` o `iloc`. Al trabajar en la respuesta de esta pregunta y de varias de las que siguen, ten en cuenta la siguiente "trampa" descrita en la [referencia](https://www.kaggle.com/residentmario/indexing-selecting-assigning-reference) de esta sección del tutorial:

> `iloc` usa el esquema de indexación de la biblioteca estándar de Python, donde el primer elemento del rango se incluye y el último se excluye. Así, `0:10` seleccionará las entradas `0,...,9`. `loc`, en cambio, indexa de forma inclusiva. Así, `0:10` seleccionará las entradas `0,...,10`.

> [...]

> ...[considera] cuando el índice del DataFrame es una lista numérica simple, por ejemplo `0,...,1000`. En este caso `reviews.iloc[0:1000]` devolverá 1000 entradas, mientras que `reviews.loc[0:1000]` devolverá 1001. Para obtener 1000 elementos usando `iloc`, tendrás que subir uno y pedir `reviews.iloc[0:1001]`.

```python
df = reviews.loc[0:99,['country','variety']]

q7.check()
df
```

Crea un DataFrame `italian_wines` que contenga reseñas de vinos elaborados en `Italy`. Pista: ¿a qué es igual `reviews.country`?

## 9.

Crea un DataFrame `top_oceania_wines` que contenga todas las reseñas con al menos 95 puntos (de 100) para vinos de Australia o Nueva Zelanda.


```python
top_oceania_wines = reviews[(reviews.points>=95) & ((reviews.country=='Australia') | (reviews.country=='New Zealand'))]


q9.check()
top_oceania_wines
```
