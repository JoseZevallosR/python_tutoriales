# Introducción
Los mapas nos permiten transformar datos en un `DataFrame` o `Series` un valor a la vez para una columna completa. Sin embargo, con frecuencia queremos agrupar nuestros datos y luego hacer algo específico con el grupo al que pertenecen. Esto lo hacemos con la operación `groupby`.

En estos ejercicios aplicaremos análisis por grupos a nuestro conjunto de datos.

## 3.
¿Cuáles son los precios mínimos y máximos para cada `variety` de vino? Crea un `DataFrame` cuyo índice sea la categoría `variety` del conjunto de datos y cuyos valores sean sus valores `min` y `max`.


```python
price_extremes = reviews.groupby('variety').price.agg([min, max])
q3.check()
```

## 4.

```python
sorted_varieties = price_extremes.sort_values(by=['min', 'max'], ascending=False)
#print(sorted_varieties.loc[:,'min']).sort_index()
q4.check()
```

## 5.
Crea una `Series` cuyo índice sean los catadores y cuyos valores sean el puntaje promedio de reseña otorgado por cada catador. Pista: necesitarás las columnas `taster_name` y `points`.


```python
reviewer_mean_ratings = reviews.groupby('taster_name').points.mean()
print(reviewer_mean_ratings)
q5.check()
```

## 6.
¿Qué combinación de países y variedades es la más común? Crea una `Series` cuyo índice sea un `MultiIndex` de pares `{country, variety}`. Por ejemplo, un pinot noir producido en EE. UU. debería corresponder a `{"US", "Pinot Noir"}`. Ordena los valores de la `Series` en orden descendente según el conteo de vinos.


```python
country_variety_counts = reviews.groupby(['country','variety']).size().sort_values(ascending=False)
print(country_variety_counts)
q6.check()
```
