```python
import pandas as pd
import numpy as np

# Cargar la base de datos
df = pd.read_csv(
    '../Data/05_PAS-MOD_TRANSFERENCIA_PNSU.csv',
    encoding='latin1',
    sep=';'
)

# Eliminar columnas auxiliares sin nombre
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

# Normalizar los nombres de las columnas
df.columns = df.columns.str.strip()

print(f"Filas: {df.shape[0]:,}")
print(f"Columnas: {df.shape[1]}")
```

```python
# Preparar las variables numéricas necesarias para el análisis
for col in ['MONTO_ACTUALIZADO _PIP', 'POBLACION_BENEFICIADA_SSP']:
    df[col] = df[col].astype(str).str.replace(',', '.', regex=False).str.strip()
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Revisión inicial de la estructura de la base
df.head()
```
