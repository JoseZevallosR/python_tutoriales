Este ejercicio es opcional. Lee los siguientes datos SQL en un DataFrame llamado `music_reviews`:

La ruta del archivo es `../input/pitchfork-data/database.sqlite`. Pista: usa la biblioteca `sqlite3`. El nombre de la tabla es `artists`.


````python
import sqlite3

conn = sqlite3.connect('Data/database.sqlite')

tables = pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

tables
````

````python
music_reviews = pd.read_sql('SELECT * FROM artists', conn)

music_reviews.head()
````
