## 1) ¿Qué es una variable?
Una **variable** es un nombre que referencia un valor en memoria.

- En Python no declaras el tipo: el tipo depende del valor asignado.
- Puedes reasignar el nombre a otro valor (y a otro tipo).

```python
# Ejemplos
x = 10
print(x, type(x))

x = "hola"
print(x, type(x))
```

## 2) Tipos básicos: string, int, float
En Python los tipos comunes son:

- `str` (string): texto
- `int`: enteros
- `float`: números reales (con decimales)

**Nota:** en Python no existe `double` como tipo separado; `float` ya usa precisión doble internamente (IEEE 754).

```python
nombre = "Ana"     # str
edad = 20         # int
altura = 1.72     # float

print(nombre, type(nombre))
print(edad, type(edad))
print(altura, type(altura))
```

### Conversión de tipos
A veces necesitas convertir:
- `int('12')` → 12
- `float('3.14')` → 3.14
- `str(10)` → '10'

```python
a = "12"
b = "3.14"

print(int(a) + 1)
print(float(b) * 2)
print(str(100) + " soles")
```

## 3) Listas
Una **lista** es una colección ordenada y *mutable*.

Operaciones frecuentes:
- Crear: `[]` o `list()`
- Indexar: `lista[0]`
- Slicing: `lista[1:3]`
- Agregar: `append`, `extend`
- Remover: `remove`, `pop`
- Longitud: `len(lista)`

```python
nums = [10, 20, 30]
print(nums[0])      # 10
print(nums[-1])     # 30

nums.append(40)
print(nums)

nums.pop()          # quita el último
print(nums)

nums2 = [1, 2]
nums.extend(nums2)
print(nums)
```

### Operaciones útiles con listas
- `sum(lista)`
- `min(lista)`, `max(lista)`
- `sorted(lista)` (no modifica)
- `lista.sort()` (modifica)

```python
vals = [3, 1, 2]
print(sorted(vals))  # [1, 2, 3]
print(vals)          # sigue igual

vals.sort()
print(vals)          # ahora sí cambia
```

## 4) Tuplas
Una **tupla** es una colección ordenada e *inmutable*.

Se usa cuando no quieres que cambie (por ejemplo, coordenadas).

```python
punto = (10, 5)
print(punto[0], punto[1])
print(type(punto))
```

## 5) Diccionarios
Un **diccionario** almacena pares **clave → valor**.

Operaciones:
- Crear: `{}`
- Acceder: `d['clave']`
- Agregar/modificar: `d['clave'] = valor`
- Recorrer: `for k, v in d.items(): ...`

```python
alumno = {"id": "A001", "nombre": "Pedro", "nota": 15}
print(alumno["nombre"])

alumno["nota"] = 18
alumno["curso"] = "Python"
print(alumno)

for k, v in alumno.items():
    print(k, "->", v)
```
