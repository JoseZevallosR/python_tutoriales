## 1. Conceptos: Clases e Instancias

Una **Clase** es un plano arquitectónico o molde. Define qué datos y qué comportamientos tendrá un objeto.

Una **Instancia** es el objeto real creado a partir de esa clase.

### El método `__init__` y `self` 
- `__init__`: Es el método **constructor**. Se ejecuta automáticamente al crear un objeto.
- `self`: Es una palabra clave que representa a la **instancia actual**. Se usa para acceder a las variables que pertenecen a la clase.

```python
class Robot:
    def __init__(self, nombre, modelo):
        # Atributos de instancia
        self.nombre = nombre
        self.modelo = modelo
        self.energia = 100

# Creación de instancias
mi_robot = Robot("Wall-E", "Compactador")
print(f"Robot: {mi_robot.nombre} | Modelo: {mi_robot.modelo}")
```

## 2. Métodos (Desde init hasta propios)

Los métodos son funciones que viven dentro de una clase y definen lo que el objeto puede **hacer**.

```python
class Robot:
    def __init__(self, nombre):
        self.nombre = nombre
        self.bateria = 100
    
    # Método propio: define una acción
    def saludar(self):
        return f"Hola, soy {self.nombre}. Mi batería está al {self.bateria}%."
    
    def trabajar(self, costo):
        self.bateria -= costo
        return f"{self.nombre} está trabajando... Batería restante: {self.bateria}%"

bot = Robot("Eva")
print(bot.saludar())
print(bot.trabajar(20))
```

## 3. Herencia

La herencia permite que una clase hija adquiera los atributos y métodos de una clase padre, evitando repetir código.

```python
class Vehiculo:
    def __init__(self, marca):
        self.marca = marca
    
    def encender(self):
        return f"El {self.marca} ha encendido."

# Drone hereda de Vehiculo
class Drone(Vehiculo):
    def volar(self):
        return f"El drone {self.marca} está despegando."

mi_drone = Drone("DJI")
print(mi_drone.encender()) # Método heredado
print(mi_drone.volar())    # Método propio
```
